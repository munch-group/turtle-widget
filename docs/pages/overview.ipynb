**turtle-widget** is an animated, [`turtle`](https://docs.python.org/3/library/turtle.html)-compatible
drawing widget for Jupyter notebooks. You write ordinary turtle code in Python; the widget
records each command and replays it as a smooth canvas animation right below the cell.

Built on [`anywidget`](https://anywidget.dev), it behaves identically in **VS Code**,
**JupyterLab**, **Notebook 7**, and **Colab**, and works offline.

````python
from turtle_widget import Turtle

t = Turtle()
t.speed(6)
for _ in range(4):     # a square
    t.forward(120)
    t.left(90)
````

## Multiple turtles

Two or more turtles can share one canvas. `new_turtle()` creates a turtle that draws on
the same canvas as an existing one — giving it a different colour makes the two easy to
tell apart — and each still animates in the order its commands run, one at a time.

````python
t1 = Turtle()
t2 = t1.new_turtle(color="red")

for _ in range(4):
    t1.forward(100)
    t1.left(90)
    t2.forward(60)
    t2.right(90)
````
